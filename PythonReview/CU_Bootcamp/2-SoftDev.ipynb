```python
### 2.2 - Comprehensions
# Append items to a list
list_ = []
for i in range(10):
    list_.append(i)
list_

# Single Line Loops
[i for i in range(10)]
[x for x in range(10) if x%2] # Conditonals can be added on

# Nested Comprehensions
# Loops work inside -> out, x held constant while y iterates to exhaustion
# then x is incremented and y loops again
[('x' + str(x), 'y' + str(y))
for x in range(3)
for y in range(2)]

# Nested Loops with Append
new_list = []
for x in range(3):
    for y in range(2):
        new_list.append(('x' + str(x), 'y' + str(y)))
print(new_list)

# Dict Comprehensions
{x: x**2 for x in range(10) if x%2}
```

```python
### 2.3 - *args and *kargs
# *args tells python to accept 0 or more args
def myfunc1(arg1, *args):
    print(arg1)
    print(*args) # additional args wrapped in tuple (iterable)
myfunc1(1, 2, 3, 4, 5) # returns 1 and then 2 3 4 5

# args is iterable
def myfunc2(*args):
    for a in args:
        print(a)
print(myfunc2(1, 2, 3, 4))

# kargs (keyword args) are used for creating dicts
def myfunc3(**kargs):
    print(kargs)
myfunc3(a=1, b=2, c=3)
```

```python
### 2.4 - Keyword Mutability
def afunc(list_):
    internal_list = list(list_) # casting list to a list makes a copy
    internal_list.append('HI')
    return internal_list

lis = []
print(afunc(lis))
print(afunc(lis))
print(afunc(lis))
```

```python
### 2.6 - Lambda Functions
# Define a lambda function that squares a number
square = lambda x: x**2  # Equivalent to: def square(x): return x**2

# Use map() to apply the lambda to each element in the list [1, 2, 3]
# map() returns a map object, so we convert it to a list
print(list(map(lambda x: x**2, [1, 2, 3])))  # Output: [1, 4, 9]

# Immediately invoke a lambda that takes any number of arguments (*args)
# and returns their sum
print((lambda *args: sum(args))(2, 2, 2))  # Output: 6

# Exercise: Convert Celsius to Fahrenheit using a lambda inside map
# Formula: F = (C * 9/5) + 32
print(list(map(lambda c: c * (9/5) + 32, [1, 2, 3])))  # Output: [33.8, 35.6, 37.4]

```
