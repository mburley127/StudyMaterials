```python
### 1.8 - Loops
# For Loops
for x in range(5):
    print(x)

for y in [1, 2, 3]:
    print(y)

# While Loops - Loop until a condition is met
i = 0
while(i<0):
    print(i)
    i+=1

# Looping over functions
def perform_op(x):
    return ((x-2)+54) / 3

for x in range(-2,2):
    print(round(perform_op(x), 3))
```

```python
### 1.9 - File I/O
# Create and Populate a .txt file
with open('open_test.txt', 'w') as fp:
    fp.write('This is my test string for this exercise')

# Open a file
with open('open_test.txt') as fp:
    content = fp.read()
content

# Add(Replace) lines in a file
lines = [
    'line 1',
    'line 2',
    'line 3',
]
with open('open_test.txt', 'w') as fp:
    content = fp.writelines(lines)

# Replace Lines in a file with newline breaks
with open('open_test.txt', 'w') as fp:
    fp.writelines(map(lambda line: line + '\n', lines))
with open('open_test.txt') as fp:
    content2 = fp.read()
print(content2)
```

```python
### 1.10 - Type Annotation
# Values passed into functions to hint at type
def my_int_sum(a: int, b: int) -> int:
    return sum(a, b)
my_int_sum(1, 3)

# Define a function that accepts any number of float arguments and returns their sum
from typing import Sequence
del sum # Reset the sum

def get_sum(*args: Sequence[float]) -> float:
    return sum(args)
print(get_sum(1.0, 2.0, 3.0))
```
