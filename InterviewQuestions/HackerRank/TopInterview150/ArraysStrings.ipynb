### Remove Element (String/Array)

Given an integer array `nums` and an integer `val`, remove all occurrences of `val` in `nums` in-place. The order of the elements may be changed. Then return the number of elements in `nums` which are not equal to `val`.

Consider the number of elements in `nums` which are not equal to `val` to be `k`. To get accepted, you need to do the following things:

- Change the array `nums` such that the first `k` elements of `nums` contain the elements which are not equal to `val`. The remaining elements of `nums` are not important, as well as the size of `nums`.
- Return `k`.

**Input**: `nums = [3,2,2,3]`, `val = 3`  
**Output**: `2`, `nums = [2,2,_,_]`


```python
### Remove Element (String/Array)
class Solution(object):
    def removeElement(self, nums, val):
        """
        :type nums: List[int]
        :type val: int
        :rtype: int
        """
        # Pointer to track the position of valid elements
        k = 0
        # Iterate through the array
        for i in range(len(nums)):
            if nums[i] != val:
                # Place the valid element at index k
                nums[k] = nums[i]
                k += 1     
        # Return the count of elements not equal to val (k)
        return k  

# Test Cases
nums = [3, 2, 2, 3]
val = 3
solution = Solution()
k = solution.removeElement(nums, val)

# Test the function and print results
print("k:", k)                      # Output: 2
print("Modified nums:", nums[:k])   # Output: [2, 2]
```

### Remove Duplicates from Sorted Array

Given an integer array `nums` sorted in non-decreasing order, remove the duplicates **in-place** such that each unique element appears only once. The relative order of the elements should be kept the same. Then return the number of unique elements in `nums`.

Consider the number of unique elements in `nums` to be `k`. To get accepted, you need to do the following things:

- Change the array `nums` such that the first `k` elements of `nums` contain the unique elements in the order they were present in `nums` initially. The remaining elements of `nums` are not important, as well as the size of `nums`.
- Return `k`.

**Input**: `nums = [1,1,2]`  
**Output**: `2`, `nums = [1,2,_]`


```python
### Remove Duplicates from Sorted Array
class Solution(object):
    def removeDuplicates(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        # Initialize two pointers
        k = 0  # Pointer to track the position of unique elements

        # Iterate through the array with pointer j
        for j in range(1, len(nums)):
            # If the current element is different from the last unique element
            if nums[j] != nums[k]:
                k += 1  # Move k to the next position
                nums[k] = nums[j]  # Update nums[k] with the new unique element

        # Return the number of unique elements
        return k + 1  # +1 because k is index-based
    
# Test cases
nums1 = [1, 1, 2]
nums2 = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]

solution = Solution()

print(solution.removeDuplicates(nums1))  # Output: 2
print(nums1[:2])  # Output: [1, 2]

print(solution.removeDuplicates(nums2))  # Output: 5
print(nums2[:5])  # Output: [0, 1, 2, 3, 4]
```

### Best Time to Buy and Sell Stock

You are given an array `prices` where `prices[i]` is the price of a given stock on the `i`th day.

You want to **maximize your profit** by choosing a **single day to buy one stock** and choosing a **different day in the future** to sell that stock.

Return the **maximum profit** you can achieve from this transaction. If you cannot achieve any profit, return `0`.


```python
### Best Time to Buy a Stock
class Solution(object):
    def maxProfit(self, prices):
        """
        :type prices: List[int]
        :rtype: int
        """
        buy = float('inf')  # Initialize buy price as infinity
        max_profit = 0  # Start with zero profit

        for price in prices:
            # Update the minimum buy price
            if price < buy:
                buy = price
            # Calculate profit if sold at current price
            elif price - buy > max_profit:
                max_profit = price - buy  # Update max profit

        return max_profit

# Test Cases
prices1 = [7, 1, 5, 3, 6, 4]
prices2 = [4, 3, 5, 1, 9]
prices3 = [7, 6, 4, 3, 1]  # No profit case

solution = Solution()
print(solution.maxProfit(prices1))  # Expected output: 5
print(solution.maxProfit(prices2))  # Expected output: 8
print(solution.maxProfit(prices3))  # Expected output: 0
```

### Length of Last Word

Given a string `s` consisting of words and spaces, return the **length of the last word** in the string.

A **word** is defined as a maximal sequence of **non-space characters**.

#### **Example 1**
**Input**: `s = "Hello World"`  
**Output**: `5`  
**Explanation**: The last word is `"World"` with length **5**.

#### **Example 2**
**Input**: `s = "   fly me   to   the moon  "`  
**Output**: `4`  
**Explanation**: The last word is `"moon"` with length **4**.


```python
### Length of Last Word
class Solution(object):
    def lengthOfLastWord(self, s):
        """
        :type s: str
        :rtype: int
        """
        # Initialize counter for the last word's length
        count = 0
        # Flag to start counting only when we hit a non-space character
        found_char = False
        # Iterate over the string from the end to the beginning
        for i in range(len(s) - 1, -1, -1):  # range(start, stop, step)
            if s[i] != ' ':
                count += 1
                found_char = True
            elif found_char:
                # If there is already found a word and hit a space, break
                break

        return count

# Test Cases
s1 = 'Hello World'
s2 = ' fly me to the moon '

solution = Solution()
print(solution.lengthOfLastWord(s1))  # Expected output: 5
print(solution.lengthOfLastWord(s2))  # Expected output: 4
```
