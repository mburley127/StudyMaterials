### 20. Valid Parentheses

**Problem**:  
Given a string `s` containing just the characters `'('`, `')'`, `'{'`, `'}'`, `'['` and `']'`, determine if the input string is valid.

An input string is valid if:

- Open brackets must be closed by the same type of brackets.  
- Open brackets must be closed in the correct order.  
- Every close bracket has a corresponding open bracket of the same type.


```python
# Valid Parentheses
class Solution(object):
    def isValid(self, s):
        """
        :type s: str
        :rtype: bool
        """
        # Initialize an empty list which acts as our stack
        stack = []
        # Dict to match closing to opening brackets
        mapping = {')': '(', ']': '[', '}': '{'}
        
        for char in s:
            if char in mapping:
                # Pop from stack if not empty, else use dummy '#'
                top = stack.pop() if stack else '#'
                if mapping[char] != top:
                    return False
            else:
                stack.append(char)                    
        return not stack
    
# Test Case
s = '[]{}()' 
t = '[}()'

solution = Solution()
print(solution.isValid(s))  # Expected output: True
print(solution.isValid(t))  # Expected output: False
```

### 58. Length of Last Word

Given a string `s` consisting of words and spaces, return the **length of the last word** in the string.

A **word** is defined as a maximal sequence of **non-space characters**.

#### **Example 1**
**Input**: `s = "Hello World"`  
**Output**: `5`  
**Explanation**: The last word is `"World"` with length **5**.

#### **Example 2**
**Input**: `s = "   fly me   to   the moon  "`  
**Output**: `4`  
**Explanation**: The last word is `"moon"` with length **4**.


```python
### Length of Last Word
class Solution(object):
    def lengthOfLastWord(self, s):
        """
        :type s: str
        :rtype: int
        """
        # Initialize counter for the last word's length
        count = 0
        # Flag to start counting only when we hit a non-space character
        found_char = False
        # Iterate over the string from the end to the beginning
        for i in range(len(s) - 1, -1, -1):  # range(start, stop, step)
            if s[i] != ' ':
                count += 1
                found_char = True
            elif found_char:
                # If there is already found a word and hit a space, break
                break

        return count

# Test Cases
s1 = 'Hello World'
s2 = ' fly me to the moon '

solution = Solution()
print(solution.lengthOfLastWord(s1))  # Expected output: 5
print(solution.lengthOfLastWord(s2))  # Expected output: 4
```
