### 169. Majority Element (String/Array)
Given an array `nums` of size `n`, return the majority element. 
The majority element is the element that appears more than ⌊n / 2⌋ times. You may assume that the majority element always exists in the array.

#### Example 1:
**Input**: `nums = [3,2,3]`  
**Output**: `3`

#### Example 2:
**Input**: `nums = [2,2,1,1,1,2,2]`  
**Output**: `2`

```python
### Majority Element
class Solution(object):
    def majorityElement(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        # Dictionary to store counts
        counts = {}
        # Iterate over the nums array
        for num in nums:
            # If num is already in the counts dict, add it
            if num in counts:
                counts[num] += 1
            else:
                counts[num] = 1
        # Return the highest value index in the array
        max_key = max(counts, key = counts.get)
        return max_key

# Test cases
nums1 = [3, 2, 3]
nums2 = [2, 2, 1, 1, 1, 2, 2]

# Instantiate the solution
solution = Solution()
# Test the function and print results
print(solution.majorityElement(nums1))  # Expected output: 3
print(solution.majorityElement(nums2))  # Expected output: 2
```

## 136. Single Number (169 but Linear!!!)

**Problem:**  
Given a non-empty array of integers `nums`, every element appears **twice** except for one. Find that single one.

You must implement a solution with **linear runtime complexity** and use only **constant extra space**.


```python
# Single Number
class Solution(object):
    def singleNumber(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        # Dictionary to store counts
        counts = {}
        # Iterate over the nums array
        for num in nums:
            # If num is already in the counts dict, add it
            if num in counts:
                counts[num] += 1
            else:
                counts[num] = 1
        # Return the highest value index in the array
        min_key = min(counts, key = counts.get)
        return min_key

# Test cases
nums1 = [3, 2, 3]
nums2 = [2, 2, 1, 1, 3, 2, 2]

# Instantiate the solution
solution = Solution()
# Test the function and print results
print(solution.singleNumber(nums1))  # Expected output: 2
print(solution.singleNumber(nums2))  # Expected output: 2
```

### Roman to Integer
#### Roman numerals are represented by seven different symbols:

| Symbol | Value |
|--------|-------|
| I      | 1     |
| V      | 5     |
| X      | 10    |
| L      | 50    |
| C      | 100   |
| D      | 500   |
| M      | 1000  |

- **Basic Addition**:  
  - `II = 1 + 1 = 2`
  - `XII = 10 + 2 = 12`
  - `XXVII = 20 + 5 + 2 = 27`

- **Subtraction Rule**:  
  Roman numerals are usually written **largest to smallest** from left to right.  
  However, there are cases where **subtraction** is used:
  - `IV = 4` (since `I` comes before `V`, subtract `1`)
  - `IX = 9` (since `I` comes before `X`, subtract `1`)

  **Instances where subtraction is used:**
  - `I` can be placed before `V (5)` and `X (10)` → **4 and 9**  
  - `X` can be placed before `L (50)` and `C (100)` → **40 and 90**  
  - `C` can be placed before `D (500)` and `M (1000)` → **400 and 900**  

Given a **Roman numeral**, convert it to an **integer**.

```python
### Roman to Integer
class Solution(object):
    def romanToInt(self, s):
        """
        :type s: str
        :rtype: int
        """
        # Dictionary to Store Roman Numerals
        roman_values = {
            'I': 1, 'V': 5, 'X': 10, 'L': 50,
            'C': 100, 'D': 500, 'M': 1000
        }
        # Store total and the previous value
        total = 0
        prev_value = 0 
        # Iterate over the string **from right to left**
        for char in reversed(s):
            current_value = roman_values[char]  # Look up the value from the dictionary
            # If the current value is smaller than prev_value, subtract it from total
            if current_value < prev_value:
                total -= current_value  # Subtract if a smaller numeral appears before a larger one
            else:
                total += current_value  # Otherwise, add the value
            # Update prev_value for the next iteration
            prev_value = current_value  

        return total

# Test Cases
s1 = 'III'
s2 = 'MCMXCIV'

solution = Solution()
print(solution.romanToInt(s1))  # Expected output: 3
print(solution.romanToInt(s2))  # Expected output: 1994
```

## 1. Two Sum

**Problem:**  
Given an array of integers `nums` and an integer `target`, return the indices of the two numbers such that they add up to `target`.

You may assume that each input would have **exactly one solution**, and you **may not use the same element twice**.

You can return the answer in **any order**.


```python
### Two Sum
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        # Create an empty dict to store values and their indices
        vals = {}
        # Iterate over the indices 
        for i in range(len(nums)):
            # Compute the complement
            complement = target - nums[i]
            # Check if complement exists in the dictionary
            if complement in vals:
                return [vals[complement], i]
            # Store the current number and its index in the dictionary
            vals[nums[i]] = i
        
        # If no solution is found, return an empty list (though the problem guarantees one)
        return []

# Test Cases
list1 = [1, 2, 4, 5]
target1 = 6
list2 = [22, 7, 9]
target2 = 29

solution = Solution()
print(solution.twoSum(list1, target1))  # Expected output: [1, 2]
print(solution.twoSum(list2, target2))  # Expected output: [0, 1]
```

## 448. Find All Numbers Disappeared in an Array

**Problem:**  
Given an array `nums` of `n` integers where `nums[i]` is in the range `[1, n]`, return an array of all the integers in the range `[1, n]` that do **not** appear in `nums`.


```python
class Solution(object):
    def findDisappearedNumbers(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        # Mark each number in the seen dict (1 if seen)
        seen = {}
        # Mark each number in nums with a 1 if seen
        for num in nums:
            seen[num] = 1

        # Check for numbers from 1 to n that are not in seen, append to result dict
        result = []
        for i in range(1, len(nums) + 1):
            if i not in seen:
                result.append(i)
        
        return result

# Test Cases
list1 = [1, 2, 4, 5]
list2 = [2, 2, 5, 4]

solution = Solution()
print(solution.findDisappearedNumbers(list1))  # Expected output: [1, 2]
print(solution.findDisappearedNumbers(list2))  # Expected output: [0, 1]
```
