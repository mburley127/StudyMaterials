```python
### Find the smallest number in a list
def smallest_number(nums):
    # Init var to store first val in nums array
    smallest = nums[0]
    
    for i in range(len(nums)):
        # If current num < smallest num, the smallest num becoms the curr num
        if nums[i] < smallest:
            smallest = nums[i]

    return smallest

# Test Case
nums = [4, 1, 5, 8, 7, 2]
print(smallest_number(nums))
```

```python
### Find Average in a list
def compute_avg(nums):
    # Init vars to store first sum and avg
    sum = 0
    count = 0

    for num in nums:
        sum += num
        count += 1

    avg = sum / count
    return avg

# Test Case
nums = [4, 1, 5, 8, 4, 2]
print(compute_avg(nums))
```

```python
### Remove None from List
def remove_none(list):
    # Init counter var
    i = 0
    # While Loop to iterate over list
    while i < len(list):
        if list[i] is None:
            del list[i]
        else:
            i += 1
    return list

# Test Case
nums = [4, None, 5, 8, None, 2]
print(remove_none(nums))
```

```python
### Intersection of Two Lists - Initial Solution
# The intersection is defined as the elements present in BOTH List A and List B
def intersection(list1, list2):
    # Initialize an empty list to store intersection elements
    int_list = []
    
    # Iterate over all elements in list1
    for i in range(len(list1)):
        # Iterate over all elements in list2
        for j in range(len(list2)):
            # If the current element in list1 matches an element in list2, add it to the intersection list
            if list1[i] == list2[j]:
                int_list.append(list1[i])
    
    return int_list

# Test Case
list_a = [1, 2, 3, 4]
list_b = [3, 4, 5, 6]
print('The nested loops together result in O(n×m) operations, as every element of list1 is compared with every element of list2')
print(intersection(list_a, list_b))  # Output: [3, 4]
```

```python
### Intersection of Two Lists - O(n)
# The intersection is defined as the elements present in BOTH List A and List B
def intersection(list1, list2):
    # Convert one of the lists to a set for O(1) lookup
    set_list2 = set(list2)
    # Use list comprehension to find elements in list1 that are also in set_list2
    int_list = [num for num in list1 if num in set_list2]
    
    return int_list

# Test Case
list_a = [1, 2, 3, 4]
list_b = [3, 4, 5, 6]
print(intersection(list_a, list_b))  # Output: [3, 4]
```
