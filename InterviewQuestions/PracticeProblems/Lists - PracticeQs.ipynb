```python
### Find the smallest number in a list
def smallest_number(nums):
    # Init var to store first val in nums array
    smallest = nums[0]
    
    for i in range(len(nums)):
        # If current num < smallest num, the smallest num becoms the curr num
        if nums[i] < smallest:
            smallest = nums[i]

    return smallest

# Test Case
nums = [4, 1, 5, 8, 7, 2]
print(smallest_number(nums))
```

```python
### Find Average in a list
def compute_avg(nums):
    # Init vars to store first sum and avg
    sum = 0
    count = 0

    for num in nums:
        sum += num
        count += 1

    avg = sum / count
    return avg

# Test Case
nums = [4, 1, 5, 8, 4, 2]
print(compute_avg(nums))
```

```python
### Remove None from List
def remove_none(list):
    # Init counter var
    i = 0
    # While Loop to iterate over list
    while i < len(list):
        if list[i] is None:
            del list[i]
        else:
            i += 1
    return list

# Test Case
nums = [4, None, 5, 8, None, 2]
print(remove_none(nums))
```

```python
### Intersection of Two Lists - Initial Solution
# The intersection is defined as the elements present in BOTH List A and List B
def intersection(list1, list2):
    # Initialize an empty list to store intersection elements
    int_list = []
    
    # Iterate over all elements in list1
    for i in range(len(list1)):
        # Iterate over all elements in list2
        for j in range(len(list2)):
            # If the current element in list1 matches an element in list2, add it to the intersection list
            if list1[i] == list2[j]:
                int_list.append(list1[i])
    
    return int_list

# Test Case
list_a = [1, 2, 3, 4]
list_b = [3, 4, 5, 6]
print('The nested loops together result in O(n×m) operations, as every element of list1 is compared with every element of list2')
print(intersection(list_a, list_b))  # Output: [3, 4]
```

```python
### Intersection of Two Lists - O(n)
# The intersection is defined as the elements present in BOTH List A and List B
def intersection(list1, list2):
    # Convert one of the lists to a set for O(1) lookup
    set_list2 = set(list2)
    # Use list comprehension to find elements in list1 that are also in set_list2
    int_list = [num for num in list1 if num in set_list2]
    
    return int_list

# Test Case
list_a = [1, 2, 3, 4]
list_b = [3, 4, 5, 6]
print(intersection(list_a, list_b))  # Output: [3, 4]
```

```python
### Reverse an Array In-Place
''' 
Write a function that takes an array as input and reverses the elements of the array in place (i.e., without using an additional array). 
You can only use a single additional variable for temporary storage.
'''
def reverse_array(arr):
    i = 0
    j = len(arr) - 1
    
    while i < j:
        # Swap elements at positions i and j
        arr[i], arr[j] = arr[j], arr[i]
        # Move the pointers
        i += 1
        j -= 1
    
    return arr

# Test Case
arr = [1, 2, 3, 4, 5]
print(reverse_array(arr))
```

```python
### Count the Occurrences of a Target Element
''' 
Given an array of integers and a target integer, write a function that returns the number of times the target appears in the array.
'''
def count_target(arr, target):
    # Count variable
    count = 0
    for v in arr:
        if v == target:
            count += 1
    
    return count

# Test Case
arr = [5, 1, 2, 5, 3, 4, 5]
print(count_target(arr, target = 5))
```

```python
### Find the Sum of the Positive Numbers
''' 
Write a function that takes a list of integers and returns the sum of all the positive numbers in the list.
'''
def pos_sum(nums):
    # Initialize sum var
    plus_sum = 0

    for num in nums:
        if num > 0:
            plus_sum += num
    
    return plus_sum

# Test Case
nums = [-1, 2, 3, -5, 6, 7]
print(pos_sum(nums))
```

```python
### Get Unique Values from a List in Python
def get_unique_vals(nums):
    # Create empty list
    unique_vals = []

    for num in nums:
        if num not in unique_vals:
            unique_vals.append(num)

    return unique_vals

nums = [1, 2, 1, 1, 3, 4, 3, 3, 5]
print(get_unique_vals(nums))
```

```python
### Find the Sum of number digits in List in Python
def sum_digits(nums):
    # Create empty list
    sum_arr = []

    for num in nums:
        # Reset sum_digits for each number
        sum_digits = 0
        for digit in str(num):
            sum_digits += int(digit)

        sum_arr.append(sum_digits)

    return sum_arr

nums = [191, 227, 58]
print(sum_digits(nums))
```

```python
### Count Vowels in a String
''' 
Write a function that takes a string as input and returns the number of vowels (a, e, i, o, u) in the string. The function should ignore case.
''' 
def count_vowels(s):
    # Init vowels and count variable
    vowels = "aeiouAEIOU"
    count = 0
    # Iterate over each char in the string
    for char in s:
        if char in vowels:
            count += 1

    return count

# Test Case
print(count_vowels("Programming is fun"))
```

```python
### Find the Longest Word in a List
''' 
Write a Python function that takes a list of words and returns the longest word. If multiple words have the same length, return the first one.
'''
def longest_word(words):
    # Set the longest word equal to the first word
    longest = words[0]
    # Iterate over each word in the words list
    for word in words:
        if len(word) > len(longest):
            longest = word
        
    return longest

# Test Case
words = ["Word", "ThisIsTheLongest", "python"]
print(longest_word(words))
```

```python
### Min/Max Elements in an Array
''' 
Given an array arr. Your task is to find the minimum and maximum elements in the array.
'''
def get_min_max(arr):
    # Init Min and Max Values
    min_val = arr[0]
    max_val = arr[0]
        
    for num in arr:
        # Calculate the Max Value
        if max_val > num:
            max_val = num
        # Calculate the Min Value
        elif min_val < num:
            min_val = num
                
    return min_val, max_val

arr = [1, 99, 8872, 4, 19]
print(get_min_max(arr))
```
